```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.io import loadmat
```

```python
NORMAL_DIR = Path("../data/raw/normal")
DE_12K_DIR = Path("../data/raw/12k_drive_end")


def inspect_mat_file(fileName, path, fs):
    mat = loadmat(path/fileName)

    print(f"Inspecting {fileName} in {path}\n")
    print(f"mat.keys(): {mat.keys()}\n")

    # keys startswith "__" are metadata, so we skip them
    for key, value in mat.items():
        if key.startswith("__"):
            continue
        print(f"{key:15s} shape={value.shape}  dtype={value.dtype}")

    # Now we look for keys that end with "DE_time".
    for key in mat.keys():
        if key.endswith("DE_time"):
            signal = mat[key]
            n = signal.shape[0]
            print(f"\n{key} first 5 values: {signal[:5].ravel()}\n")
            print(f"{n} readings / {fs} Hz = {n / fs:.2f} s")
```

```python
inspect_mat_file("Normal_0__97.mat", NORMAL_DIR, 48000)
print("\n\n")
inspect_mat_file("IR007_0__105.mat", DE_12K_DIR, 12000)
```

## Findings

- `Normal_0__97.mat` contains no `BA_time` compartment, whereas `IR007_0__105.mat` does.
- The normal file lasts approximately 5 seconds; the fault file lasts approximately 10 seconds, or about 303 shaft revolutions.
- Sampling rates: 48,000 Hz for the normal file and 12,000 Hz for the fault file.
